### Configuration

Shuffling the dataset (with a seed for reproducibility) ensures that each class label is shuffled before being scattered across train/val/test sets.

```python
input_folder = '/content/input_folder/...'
output_folder = '/content/output_folder/...'

split_ratio = (0.8, 0.1, 0.1)

splitfolders.ratio(
    input_folder, 
    output=output_folder, 
    seed=42, 
    ratio=split_ratio, 
    group_prefix=None, 
    move=False # copy the images. if move=True, delete the images from input_folder
)
```
